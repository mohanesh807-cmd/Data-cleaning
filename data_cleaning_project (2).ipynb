```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Name": ["Arun", "Bala", "Chitra", "Divya", "Arun"],
    "Math": [78, 90, 65, 88, 78],
    "Science": [85, None, 70, 92, 85],
    "English": [88, 76, 72, 95, 88],
    "Attendance": [90, 85, None, 98, 90]
}

df = pd.DataFrame(data)
df
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
sns.boxplot(data=df)
plt.title("Boxplot for Outlier Detection")
plt.show()
```

```python
df.plot(x="Name", y="Math", kind="bar")
plt.title("Math Scores")
plt.show()
```

```python
plt.hist(df["Math"])
plt.title("Distribution of Math Scores")
plt.show()
```

PROJECT INSIGHTS

1. Missing values were handled using mean method.
2. Duplicate records were removed.
3. Math and Science have positive correlation.
4. Attendance average is above 85%.
5. No major outliers detected.
